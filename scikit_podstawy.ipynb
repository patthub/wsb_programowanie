```python
# ============================================================
# scikit-learn — wprowadzenie
# ============================================================
# scikit-learn (sklearn) to biblioteka która dostarcza gotowe
# implementacje setek algorytmów ML.
#
# Nie importujemy całej biblioteki — tylko to czego potrzebujemy:
#
from sklearn.linear_model   import LinearRegression, LogisticRegression
from sklearn.tree           import DecisionTreeClassifier
from sklearn.ensemble       import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing  import StandardScaler
from sklearn.metrics        import accuracy_score, r2_score
from sklearn.datasets       import load_iris, fetch_california_housing
```

```python
# ============================================================
# GŁÓWNE MODUŁY
# ============================================================

# sklearn.linear_model  → modele liniowe
#   LinearRegression, LogisticRegression, Ridge, Lasso

# sklearn.tree          → drzewa decyzyjne -> dzis
#   DecisionTreeClassifier, DecisionTreeRegressor

# sklearn.ensemble      → modele zespołowe        -> następne zajęcia
#   RandomForestClassifier, GradientBoostingClassifier

# sklearn.svm           → maszyny wektorów nośnych
#   SVC, SVR

# sklearn.cluster       → uczenie nienadzorowane
#   KMeans, DBSCAN

# sklearn.preprocessing → przetwarzanie danych
#   StandardScaler, LabelEncoder, OneHotEncoder

# sklearn.model_selection → podział i walidacja
#   train_test_split, cross_val_score, GridSearchCV

# sklearn.metrics       → ocena modeli
#   accuracy_score, r2_score, confusion_matrix, classification_report

# sklearn.datasets      → gotowe datasety do nauki
#   load_iris, load_breast_cancer, fetch_california_housing

# sklearn.pipeline      → łączenie kroków w pipeline
#   Pipeline

# sklearn.feature_extraction → cechy z tekstu    ← przyda się przy NLP
#   TfidfVectorizer, CountVectorizer

```

```python
# ============================================================
# KLUCZOWA KONCEPCJA — jednolity interfejs
# ============================================================
# Każdy model w sklearn działa tak samo.
# Niezależnie czy to regresja, drzewo, czy random forest —
# zawsze te same trzy kroki:

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier
import numpy as np

# --- przykład z danymi ---
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 5, 4, 5])

# krok 1 — stwórz model z hiperparametrami
model = LinearRegression()

# krok 2 — wytrenuj na danych
model.fit(X, y)

# krok 3 — przewiduj na nowych danych
y_pred = model.predict([[6], [7]])
print(f"Predykcja dla 6 i 7: {y_pred}")

#sklearn wymusza żeby każdy model miał fit() i predict()
```

```python
# ============================================================
# KLUCZOWE POJĘCIA
# ============================================================

# FEATURE (cecha, X)
# -------------------------
# dane wejściowe — to co model "widzi"
# zawsze macierz: wiersze = próbki, kolumny = cechy
# przykład: wiek, dochód, lokalizacja przy przewidywaniu ceny mieszkania

# TARGET (etykieta, y)
# -------------------------
# to co chcemy przewidzieć — zawsze wektor
# regresja → liczba ciągła (cena, temperatura)
# klasyfikacja → klasa (złośliwy/łagodny, spam/nie-spam)

# HIPERPARAMETR
# -------------------------
# parametr który ustawiamy PRZED treningiem
# model go nie uczy się sam — my go wybieramy
# przykład: max_depth=5 w drzewie decyzyjnym
#
# odróżnij od parametrów modelu (np. współczynniki regresji)
# te model ustala sam podczas fit()

# GENERALIZACJA
# -------------------------
# zdolność modelu do poprawnego działania na danych których nie widział
# dobry model generalizuje — nie zapamiętuje danych treningowych
# zły model — overfitting

```

```python
# ============================================================
# TRAIN / TEST SPLIT
# ============================================================
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
import pandas as pd

cancer = load_breast_cancer()
X = pd.DataFrame(cancer.data, columns=cancer.feature_names)
y = cancer.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,    # 20% na test, 80% na trening
    random_state=42   # ziarno losowości — powtarzalność wyników
)

print(f"Wszystkie dane:  {X.shape}")
print(f"X_train:         {X_train.shape}  — cechy treningowe")
print(f"y_train:         {y_train.shape}   — etykiety treningowe")
print(f"X_test:          {X_test.shape}   — cechy testowe")
print(f"y_test:          {y_test.shape}    — etykiety testowe")
print()
print("Model oceniamy zawsze na danych których nie widział podczas treningu.")
print("Inaczej wynik byłby sztucznie zawyżony.")
```

```python
# ============================================================
# PREPROCESSING — StandardScaler
# ============================================================
from sklearn.preprocessing import StandardScaler

# Problem: cechy mają różne zakresy wartości
# jedna cecha: 0.001 - 0.5
# inna cecha:  100   - 10000
# model myśli że druga jest ważniejsza — bo większe liczby

# StandardScaler sprowadza każdą cechę do:
# średnia = 0, odchylenie standardowe = 1

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)  # ucz się średniej i odchylenia + transformuj
X_test_scaled  = scaler.transform(X_test)        # użyj tej samej średniej i odchylenia!

print(f"Przed skalowaniem — mean: {X_train['mean radius'].mean():.2f}, std: {X_train['mean radius'].std():.2f}")
print(f"Po skalowaniu    — mean: {X_train_scaled[:, 0].mean():.2f}, std: {X_train_scaled[:, 0].std():.2f}")
print()
print("WAŻNE: fit tylko na train, transform na test")
print("Naruszenie tej zasady = data leakage — wyniki sztucznie zawyżone")
```

```python
# ============================================================
# KIEDY SKALOWAĆ?
# ============================================================

# TAK — modele oparte na odległościach i gradientach:
#   LogisticRegression  ✓
#   SVM                 ✓
#   KNeighbors          ✓
#   sieci neuronowe     ✓

# NIE — modele oparte na progach:
#   DecisionTree        ✗  (pyta "czy cecha > próg", skala nie ma znaczenia)
#   RandomForest        ✗
#   GradientBoosting    ✗

#"Reguła: jeśli model mierzy odległości → skaluj. Jeśli opiera się na progach → nie trzeba
```
